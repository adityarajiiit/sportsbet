```python
import sys
import os
sys.path.insert(0,os.path.abspath('..'))
```

```python
from qdrant_client import AsyncQdrantClient
from config.settings import settings
import logging
```

```python
logger=logging.getLogger(__name__)
_client:AsyncQdrantClient|None=None
```

```python
async def newQdrant():
    global _client
    try:
        _client=AsyncQdrantClient(host=settings.QDRANT_HOST,port=settings.QDRANT_PORT)
        collections=await _client.get_collections()
        logger.info(f"connected to Qdrant collections: {collections}")
    except Exception as e:
        logger.error(f"qdrant error: {e}")
        _client=None
```

```python
def getQdrantClient()->AsyncQdrantClient|None:
    return _client
```

```python
async def closeQdrantClient():
    global _client
    if _client is not None:
        await _client.close()
```
